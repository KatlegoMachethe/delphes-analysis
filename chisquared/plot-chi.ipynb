```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Update runtime configurations globally
plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.rcParams['xtick.top'] = True
plt.rcParams['ytick.right'] = True
```

# load the chi scan data.

```python
# mass = [10, 20, 40, 60, 80, 100]
mass = [10, 50]
observable = ["dphi_a1_j1", "dphi_j1_j2", "dphi_a1_j2", "dphi_a1_bbar", "dphi_a2_bbar", "dphi_j1_bbar", "cost_bbar"]

obs_per_mass = []
for obs in observable:
    points = []
    
    for m in mass:
        scan = 'm' + str(m) + '/' + obs + '_m' + str(m) + '.txt'

        max_bound = max(pd.read_csv(scan, sep='\t')["ctt"])
        points.append(max_bound)
        # print(max_bound)
    
    obs_per_mass.append(points)
    # print(obs_per_mass)
# print(obs_per_mass)
```

#  Now plotting

```python
plt.plot(mass, obs_per_mass[0], marker='o', linestyle='--', linewidth=1, label=r'$\Delta \phi_{\gamma_1 j_1}$', color='b', markersize=5)
plt.plot(mass, obs_per_mass[1], marker='s', linestyle='--', linewidth=1, label=r'$\Delta \phi_{j_1 j_2}$', color='r', markersize=5)
plt.plot(mass, obs_per_mass[2], marker='^', linestyle='--', linewidth=1, label=r'$\Delta \phi_{\gamma_1 j_2}$', color='g', markersize=5)
plt.plot(mass, obs_per_mass[3], marker='v', linestyle='--', linewidth=1, label=r'$\Delta \phi_{\gamma_1 \bar{b}}$', color='orange', markersize=5)
plt.plot(mass, obs_per_mass[4], marker='d', linestyle='--', linewidth=1, label=r'$\Delta \phi_{\gamma_2 \bar{b}}$', color='black', markersize=5)
plt.plot(mass, obs_per_mass[5], marker='D', linestyle='--', linewidth=1, label=r'$\Delta \phi_{j_1 \bar{b}}$', color='cyan', markersize=5)
plt.plot(mass, obs_per_mass[6], marker='s', linestyle='-', linewidth=1, label=r'$\cos \theta_{\bar{b}}$', color='purple', markersize=5)

# plt.text(75, 0.725, r'$\sqrt{s} $ = 1.3 TeV', fontweight='bold', fontsize=12)
# plt.text(75, 0.7, r'$\delta_{s} $ = 5%', fontweight='bold', fontsize=12)

plt.text(10, 0.525, r'$\sqrt{s} $ = 1.3 TeV', fontweight='bold', fontsize=12)
plt.text(10, 0.5, r'$\delta_{s} $ = 5%', fontweight='bold', fontsize=12)

plt.minorticks_on()
plt.legend(frameon=False, ncol=2)
plt.title(r'Multi-bin, $\mathcal{L}$ = 1 ab$^{-1}$')
plt.xlabel(r'$m_{a}$ (GeV)', fontsize=12)
plt.ylabel(r'$C_{tt}$    (95% C.L.)', fontsize=12)

plt.savefig('chisquared.png')
plt.show()
```
