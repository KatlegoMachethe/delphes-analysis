```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

# Initializing histograms

```python
lumi = 1000
mass = ["10", "50", "100"]
observable = ["dphi_a1_j1", "dphi_j1_j2", "dphi_a1_j2", "dphi_a1_bbar", "dphi_a2_bbar", "dphi_j1_bbar", "cost_bbar"]

for m in mass:
    os.makedirs('m' + m, exist_ok=True)
    for obs in observable:
        pathSig = '../histograms/m' + m + '/' + obs + '_m' + m + '.txt'
        # pathBkg = '../histograms/m' + m + '/' + obs + '_m' + m + '_background.txt'
        pathBkg = '../histograms/m' + m + '/background/' + obs + '_m' + m + '.txt'

        sig = pd.read_csv(pathSig, sep='\t')["x"]
        bkg = pd.read_csv(pathBkg, sep='\t')["x"]
        
        # print(bkg)
        

        # calculating the error in the bins
        statErr = np.sqrt(bkg / lumi)
        sysErr  = 0.05 * bkg
        totErr  = np.sqrt(statErr**2 + sysErr**2)

        # Now doing the chi-squared scan and writing into a file
        chisquared = []
        outFile    = 'm' + m + '/' + obs + '_m' + m + '.txt'
        
        with open(outFile, 'w') as file:
            file.write('ctt\tchi2\n')
            
            for ctt in np.arange(0, 5.001, 0.001):
                chi = (sig * ctt**2) / totErr
                chi2 = sum(chi**2)
    
                if chi2 <= 3.84:
                    file.write(f'{ctt}\t{chi2}\n')
                    # print(f'{cww}\t{chi2}')

print(r'chi-squared scan complete.')
```
